```python
import numpy as np 
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.datasets import load_diabetes
```

```python
x,y=load_diabetes(return_X_y=True)
```

```python
x_train,x_test,y_train,y_test=train_test_split(x,y,test_size=0.2,random_state=2)
```

```python
class MLR:
    def __init__(self):
        self.coffecient=0
        self.intercept=0

    def fit(self,x,y):
        x=np.insert(x,0,1,axis=1)
        betas=np.linalg.inverse(np.dot(x.T,x)).dot(x.T).dot(y)
        self.coffecient=betas[1:]
        self.intercept=betas[0]

    def predict(self,x):
        return np.dot(self.coffecient,x) + self.intercept

    def display_parameters(self):
        print("cofficients: ",self.coffecient)
        print("intercpet: ",self.intercept)
```

```python
lr=MLR()
```

```python
lr.fit(x_train ,y_train)
```

```python
lr.predict(x_test)
```

```python
lr.display_parameters()
```
