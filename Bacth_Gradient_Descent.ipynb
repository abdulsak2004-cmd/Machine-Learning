```python
class BGD:
    def __init__(self,epochs=10000,learning_rate=0.01):
        self.epochs=epochs
        self.learning_rate=learning_rate
        self.coff=None
        self.intcpt=None

    def fit(self,X_train, y_train):
        self.intcpt=0
        self.coff=np.zeros(X_train.shape[1])
        for i in range(self.epochs):
            y_hat=np.dot(X_train,self.coff) + self.intcpt
            intcpt_dev = -2*np.mean(y_train-y_hat)
            coff_dev = -2*np.dot((y_train-y_hat),X_train)/(X_train.shape[0])
            self.intcpt=self.intcpt - (self.learning_rate*intcpt_dev)
            self.coff=self.coff - (self.learning_rate*coff_dev)
            if i % 1000 == 0:
                print(i, self.intcpt)

        print(self.intcpt,self.coff)
    def predict(self,x_test):
        return np.dot(X_test,self.coff)+self.intcpt
```

```python
bg=BGD()
bg.fit(X_train , y_train)
y_pred=bg.predict(X_test)
```

```python
r2_score(y_test,y_pred)
```
