```python
import numpy as np 
import pandas as pd
import matplotlib.pyplot as plt

data=pd.read_csv("placement.csv")
data.head(12)
```

```python
x=data[['cgpa']]
y=data['package']
```

```python
plt.scatter(x,y)
plt.xlabel('cgpa')
plt.ylabel('package')
```

```python
from sklearn.model_selection import train_test_split
x_train,x_test,y_train,y_test=train_test_split(x_train,y_train,test_size=0.2,random_state=2)
```

```python
x_train
```

```python
from sklearn.linear_model import LinearRegression
```

```python
lr=LinearRegression()
```

```python
lr.fit(x_train,y_train)
```

```python
x_test.iloc[[0]]
```

```python
lr.predict(x_test.iloc[[0]])
```

```python
plt.scatter(x,y)
plt.plot(x_train,lr.predict(x_train),color='red')
plt.xlabel('cgpa')
plt.ylabel('package')
```

```python
m=lr.coef_
```

```python
b=lr.intercept_
```

```python
# y= mx + b
x=8.5
m*x +b
```
