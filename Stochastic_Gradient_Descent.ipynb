```python
import numpy as np 
import pandas as pd
import matplotlib.pyplot as plt
```

```python
from sklearn.datasets import load_diabetes
from sklearn.model_selection import train_test_split
from sklearn.linear_model import SGDRegressor
from sklearn.metrics import r2_score
import random
```

```python
X,y=load_diabetes(return_X_y=True)
```

```python
X_train,X_test,y_train,y_test=train_test_split(X,y,test_size=0.2,random_state=2)
```

```python
model= SGDRegressor(max_iter=1000,learning_rate='constant',eta0=0.01)
```

```python
model.fit(X_train,y_train)
```

```python
model.intercept_
```

```python
sk_pred=model.predict(X_test)
```

```python
class SGD:
    def __init__(self,epochs=1000,learning_rate=0.01):
        self.epochs=epochs
        self.learning_rate=learning_rate
        self.coff=None
        self.intcpt=None

    def fit(self,X_train, y_train):
        self.intcpt=0
        self.coff=np.ones(X_train.shape[1])
        for i in range(self.epochs):
            for j in range(0,X_train.shape[0]):
                idx=random.randint(0,X_train.shape[0]-1)
                y_hat=np.dot(X_train[idx],self.coff) + self.intcpt
                intcpt_dev = -2*(y_train[idx]-y_hat)
                coff_dev = -2*np.dot((y_train[idx]-y_hat),X_train[idx])
                self.intcpt=self.intcpt - (self.learning_rate*intcpt_dev)
                self.coff=self.coff - (self.learning_rate*coff_dev)
                
        print(self.intcpt,self.coff)
    def predict(self,x_test):
        return np.dot(x_test,self.coff)+self.intcpt
```

```python
sgd=SGD()
```

```python
sgd.fit(X_train,y_train)
```

```python
y_pred=sgd.predict(X_test)
```

```python
from sklearn.metrics import mean_squared_error
print("My MSE:", mean_squared_error(y_test, y_pred))
print("Sklearn MSE:", mean_squared_error(y_test, sk_pred))
```
