```python
# H0=PClass and Survived are independent Columns
```

```python
import pandas as pd
import numpy as np
from scipy.stats import chi2
from scipy.stats import chi2_contingency
# Load the Titanic dataset
url = "https://raw.githubusercontent.com/datasciencedojo/datasets/master/titanic.csv"
data = pd.read_csv(url)
data.head()
```

```python
df=pd.crosstab(data['Survived'],data['Pclass'],margins=True, margins_name='Total')
df_copy=df.copy()
df
```

```python
df_copy
```

```python
Grand_Total=df.loc["Total","Total"]
for i in range(df.shape[0]-1):
    for j in range(1,df.shape[1]):
      df_copy.loc[i,j]=(df.loc[i,'Total']*df.loc['Total',j]//Grand_Total)
df_copy
```

```python
X=0
for i in range(df.shape[0]-1):
    for j in range(1,df.shape[1]):
      X+=(df.loc[i,j]-df_copy.loc[i,j])**2/df_copy.loc[i,j]
X
```

```python
degree_of_freedom=df.shape[1]-2
p_value=chi2.sf(X,df=degree_of_freedom)
p_value
```

```python
alpha=0.05
if(p_value<alpha):
    print('Reject the Null Hypothesis')
else:
     print('Failed to Reject the Null Hypothesis')
```

```python
contingency_table = pd.crosstab(data['Survived'], data['Pclass'])
contingency_table
```

```python
# Using chi2_contingency
```

```python
chi2, p_value, dof, expected = chi2_contingency(contingency_table)

print("\nChi-Square Statistic: {:.2f}".format(chi2))
print("P-value: {:.4f}".format(p_value))
print("Degrees of Freedom: {}".format(dof))
print("Expected Frequencies: \n{}".format(expected))
```

```python
alpha=0.05
if(p_value<alpha):
    print('Reject the Null Hypothesis')
else:
     print('Failed to Reject the Null Hypothesis')
```
