### `Task` Train a KNN model on glass type dataset and find best n_neighnours.

Data Link: https://drive.google.com/file/d/17cbDNBmys04MJqQfrma3jd72VPMnxIq0/view?usp=share_link

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.preprocessing import StandardScaler
from sklearn.model_selection import train_test_split
from sklearn.neighbors import KNeighborsClassifier
from sklearn.metrics import accuracy_score
```

```python
df=pd.read_csv("/content/glass.csv")
df.drop_duplicates(inplace=True)
```

```python
X=df.iloc[:,0:-1].values
y=df.iloc[:,-1]
```

```python
X_train,X_test,y_train,y_test=train_test_split(X,y,test_size=0.2,random_state=20)
```

```python
scaler=StandardScaler()
```

```python
X_train=scaler.fit_transform(X_train)
X_test=scaler.transform(X_test)
```

```python
acc_score=[]
for i in range(1,25):
  knn=KNeighborsClassifier(i)
  knn.fit(X_train,y_train)
  y_pred=knn.predict(X_test)
  print(accuracy_score(y_pred,y_test))
  acc_score.append(accuracy_score(y_pred,y_test))

plt.plot(range(1,25),acc_score,marker='o')
```

```python
# So the best accuracy score i am getting is at K=4
```
